# Как проверить

1. Выполнить restart runtime
2. Выполнить run all
3. Убедиться, что все ячейки выполнились в правильном порядке (сверху вниз) без ошибок
4. Проверить вывод, совпадает ли он с ожидаемым
5. Если возникла ошибка NameError, значит ячейка с созданием переменной и присваиванием ей значения не была выполнена до её вывода - проверить порядок ячеек
